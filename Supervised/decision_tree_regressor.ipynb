```python
#Decision Tree Regressor
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
```

```python
df=load_diabetes(as_frame=True).frame
```

```python
df.head()
```

```python
X=df.drop("target",axis=1)
y=df["target"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=0.3,random_state=42
)
```

```python
from sklearn.tree import DecisionTreeRegressor          # Import Decision Tree Regressor

model = DecisionTreeRegressor(max_depth=7,min_samples_leaf=20)                         # Create the regression model
model.fit(X_train, y_train)                              # Train the model using training data

from sklearn.metrics import r2_score, mean_squared_error # Import evaluation metrics

y_pred_train = model.predict(X_train)                   # Predict values for training data
y_pred_test = model.predict(X_test)                     # Predict values for test data

print("MSE train: ", mean_squared_error(y_train, y_pred_train))  # Calculate training MSE
print("MSE test: ", mean_squared_error(y_test, y_pred_test))     # Calculate test MSE

print("r^2 train: ", r2_score(y_train, y_pred_train))            # Calculate training R²
print("r^2 test: ", r2_score(y_test, y_pred_test))               # Calculate test R²
```

```python
from sklearn.tree import plot_tree                  # Import function to visualize the tree

plt.figure(figsize=(18, 10))                        # Set figure size

plot_tree(
    model,                                           # Trained Decision Tree Regressor
    feature_names=X.columns,                         # Display feature names
    filled=True,                                      # Add colors to the tree nodes
    
)

plt.tight_layout()                                   # Adjust spacing
plt.show()                                           # Display the Decision Tree
```
