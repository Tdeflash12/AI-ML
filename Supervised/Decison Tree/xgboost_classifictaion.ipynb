```python
import xgboost as xgb
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
```

```python
X, y = make_classification(
    n_samples=500,       # Total number of samples
    n_features=20,       # Total number of features
    n_informative=10,    # Number of informative features
    n_redundant=2,       # Number of redundant features
    random_state=42      # Ensure reproducible results
)
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y,                # Features and target labels
    test_size=0.3,       # 30% testing data and 70% training data
    random_state=42      # Ensure the same split each time
)
```

```python
xgb_clf = xgb.XGBClassifier(
    n_estimators=100,       # Number of boosting trees
    max_depth=3,             # Maximum depth of each tree
    learning_rate=0.1,       # Step size for boosting
    eval_metric="logloss",   # Evaluation metric for binary classification
    random_state=42          # Ensures reproducible results
)

# Train the model
xgb_clf.fit(X_train, y_train)

# Predict on test data
y_pred = xgb_clf.predict(X_test)

# Evaluate accuracy
from sklearn.metrics import accuracy_score

print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Display precision, recall, F1-score, and support
print(classification_report(y_test, y_pred))
```
