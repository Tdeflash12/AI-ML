```python
import pandas as pd
import matplotlib as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
iris=load_iris()
X=pd.DataFrame(iris.data)
y=iris.target
```

```python
X.head()
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
X_scaled
```

```python
#PCA
from sklearn.decomposition import PCA
pca=PCA(
    n_components=2
)
X_pca=pca.fit_transform(X_scaled)
```

```python
X_pca.shape
```

```python
print("explained variance ratio:",pca.explained_variance_ratio_)
```

```python
print("eigen vector values:",pca.components_)
```

```python
import matplotlib.pyplot as plt  # Import pyplot from Matplotlib

plt.figure(figsize=(8, 6))  # Set figure size

plt.scatter(
    X_pca[:, 0],  # Select first principal component (PCA1)
    X_pca[:, 1],   # Select second principal component (PCA2)
    c=y
)

plt.xlabel("PCA1")  # Label x-axis
plt.ylabel("PCA2")  # Label y-axis
plt.title("PCA for Iris Dataset")  # Set plot title

plt.show()  # Display the plot
```
