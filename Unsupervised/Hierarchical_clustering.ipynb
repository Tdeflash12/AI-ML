```python
from sklearn.datasets import load_iris
import matplotlib as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris=load_iris()
X=iris.data
y=iris.target
```

```python
# Scaler
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
# visualize the data
sns.scatterplot(x=X_scaled[:, 0],y=X_scaled[:,2])
```

```python
from scipy.cluster.hierarchy import linkage, dendrogram  # Import hierarchical clustering functions
# Calculate hierarchical clusters using Ward's method
Z = linkage(X_scaled, method="ward")
# Plot the dendrogram
plt.figure(figsize=(12, 6))  # Set figure size
dendrogram(Z)  # Draw the dendrogram
plt.xlabel("Samples")  # Label the x-axis
plt.ylabel("Distance")  # Label the y-axis
plt.title("Dendrogram for Hierarchical Clustering")  # Set plot title
plt.show()  # Display the plot
```

```python
# Clustering
from sklearn.cluster import AgglomerativeClustering
agg=AgglomerativeClustering(n_clusters=3)
labels=agg.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:, 0],y=X_scaled[:,2],c=labels)
```
