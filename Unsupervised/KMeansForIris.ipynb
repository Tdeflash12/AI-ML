```python
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd
```

```python
iris=load_iris()
```

```python
X=iris.data
y=iris.target
```

```python
X.shape
```

```python
y
```

```python
# Visulatize the data
sns.scatterplot(x=X[:,0],y=X[:,2],c=y)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(X)
```

```python
#optimal- dimensionality reduction using PCA
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
pca_data=pca.fit_transform(x_scaled)
```

```python
# Elbow Method
wcss = []  # Store WCSS values for each number of clusters

for k in range(1, 11):  # Test k values from 1 to 10
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)  # Create KMeans model
    kmeans.fit(pca_data)  # Fit the model on scaled data
    wcss.append(kmeans.inertia_)  # Store the WCSS value
```

```python
sns.lineplot(x=range(1,11),y=wcss,marker="o")
```

```python
# KMeans
kmeans=KMeans(n_clusters=3,random_state=10)
labels=kmeans.fit_predict(pca_data)
sns.scatterplot(x=pca_data[:,0],y=pca_data[:,1],c=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:,0],y=kmeans.cluster_centers_[:,1],marker="X",c="red",s=100)
```
