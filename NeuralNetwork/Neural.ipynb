```python
import torch
import torch.nn as nn
```

```python
#input 
torch.manual_seed(43)
inputs=torch.tensor([1.0,2.0,3.0])
```

```python
# Create a single neuron (Linear Layer)
# in_features=3 means it accepts 3 input values
# out_features=1 means it produces 1 output
neuron=nn.Linear(in_features=3,out_features=1)
```

```python
# Pass the input values through the neuron
# PyTorch calculates the weighted sum and adds bias
output=neuron(inputs)
```

```python
output
```

```python
# verify .. formula
neuron.weight @inputs + neuron.bias
```
