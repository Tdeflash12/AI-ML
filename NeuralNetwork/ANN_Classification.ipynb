```python
df=pd.read_csv("10 DateFruit_Dataset.csv")
```

```python
X=df.drop("Class",axis=1)
y=df["Class"]
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder  # Import feature scaling and label encoding

le = LabelEncoder()  # Create LabelEncoder object

y = le.fit_transform(y)  # Convert categorical labels into numerical values (0, 1, 2, ...)
```

```python
y
```

```python
from sklearn.model_selection import train_test_split  # Import train_test_split to divide dataset

X_train, X_test, y_train, y_test = train_test_split(
    X, y,                    # Input features (X) and target labels (y)
    test_size=0.2,           # Allocate 20% data for testing and 80% for training
    random_state=42          # Keep the same random split every time
)
```

```python
scaler = StandardScaler()  # Create StandardScaler object for feature scaling

X_train_scaled = scaler.fit_transform(X_train)  # Calculate mean and standard deviation from training data, then scale it

X_test_scaled = scaler.transform(X_test)  # Scale test data using training data's mean and standard deviation
```

ANN

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader,TensorDataset
```

```python
# Convert training data into PyTorch tensors

X_train_tensor = torch.tensor(
    X_train_scaled, dtype=torch.float32
)  # Convert scaled training features into FloatTensor

y_train_tensor = torch.tensor(
    y_train, dtype=torch.long
).view(-1, 1)  # Convert training labels into LongTensor and reshape into a column

# Convert testing data into PyTorch tensors
X_test_tensor = torch.tensor(
    X_test_scaled, dtype=torch.float32
)  # Convert scaled testing features into FloatTensor

y_test_tensor = torch.tensor(
    y_test, dtype=torch.long
).view(-1, 1)  # Convert testing labels into LongTensor and reshape into a column
```

```python
train_dataset = TensorDataset(
    X_train_tensor, y_train_tensor
)  # Combine training features (X) and labels (y) into one dataset

test_dataset = TensorDataset(
    X_test_tensor, y_test_tensor
)  # Combine testing features (X) and labels (y) into one dataset
```

```python
train_loader = DataLoader(
    train_dataset,
    batch_size=32,  # Process 32 training samples at a time
    shuffle=True    # Shuffle training data after each epoch
)

test_loader = DataLoader(
    test_dataset,
    batch_size=32   # Process 32 testing samples at a time
)  # No shuffling needed for evaluation
```

```python
# BUILD OUR MODEL
class ANN(nn.Module):  # Define an Artificial Neural Network class

    def __init__(self):  # Constructor to initialize the network
        super(ANN, self).__init__()  # Initialize the parent nn.Module class

        self.model = nn.Sequential(  # Combine all layers sequentially

            # 1st hidden layer
            nn.Linear(X.shape[1], 64),  # Input features → 64 neurons
            nn.ReLU(),                 # Apply ReLU activation function

            # 2nd hidden layer
            nn.Linear(64, 64),         # 64 neurons → 64 neurons
            nn.ReLU(),                 # Apply ReLU activation function

            # Output layer
            nn.Linear(64, 7),          # 64 neurons → 7 output classes
        )

    def forward(self, x):  # Define forward propagation
        return self.model(x)  # Pass input through all layers and return output
```

```python
model = ANN()

# loss, optimizer
criteria = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters())
```

```python
# Train the ANN

epochs = 100  # Number of times the model sees the entire training dataset

for epoch in range(epochs):

    model.train()  # Set model to training mode

    running_loss = 0.0  # Total training loss for one epoch

    for xb, yb in train_loader:

        # xb = Features of one batch
        # yb = Labels of one batch

        optimizer.zero_grad()  # Clear previous gradients

        outputs = model(xb)  # Forward propagation: Generate predictions

        loss = criteria(outputs, yb.squeeze(1).long())  # Calculate prediction error

        loss.backward()  # Backpropagation: Calculate gradients

        optimizer.step()  # Update model parameters (weights and biases)

        running_loss += loss.item()  # Add batch loss to total loss

    train_loss = running_loss / len(train_loader)  # Average loss per epoch

    print(f"Epoch [{epoch+1}/{epochs}], Loss: {train_loss:.4f}")
```

```python
# Evaluation
model.eval()                       # Set model to evaluation mode

total = 0                          # Initialize total test samples
correct = 0                        # Initialize correct predictions

with torch.no_grad():              # Disable gradient calculation
    for xb, yb in test_loader:     # Iterate through test batches

        outputs = model(xb)        # Pass input data through the model

        _, predicted = torch.max(outputs, 1)  # Get predicted class with highest score

        yb = yb.view(-1).long()    # Convert actual labels into 1D LongTensor
        predicted = predicted.view(-1)  # Ensure predictions are 1D

        correct += (predicted == yb).sum().item()  # Count correct predictions
        total += yb.size(0)                       # Count total samples in batch

accuracy = (correct / total) * 100  # Calculate accuracy percentage

print("Total Values:", total)       # Display total test samples
print("Correct Predictions:", correct)  # Display correct predictions
print(f"Test Accuracy: {accuracy:.2f}%")  # Display accuracy up to 2 decimal places
```
